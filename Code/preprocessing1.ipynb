```python
reviews = pd.read_csv("reviews.csv")
```

```python
reviews = reviews.rename({'overall': 'rating', 'asin': 'productID'}, axis=1)
```

```python
reviews.to_csv('reviews.csv', index=False)
```

```python
reviews
```

```python
reviews['Size'] = 0
reviews['Color'] = 0
```

```python
reviews
```

```python
for i,j in reviews.iterrows():
    x = reviews.iloc[i,4]
    if (type(x) != str) or ('Size:' not in x) or ('Color:' not in x):
        continue
    else:
    
        json_acceptable_string = x.replace("'", "\"")
        y = json.loads(json_acceptable_string)


        Size = y['Size:'].strip()
        Color = y['Color:'].strip()
        reviews.iloc[i,9] = Size
        reviews.iloc[i,10] = Color
    
```
